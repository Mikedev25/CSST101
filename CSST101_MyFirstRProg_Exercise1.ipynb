3. Numerical Operators

```r
x = 5
y = 3

add = x + y
sub = x - y
mul = x * y
div = x / y
mod = x % y

print(f"+ {add}")
print(f"- {sub}")
print(f"* {mul}")
print(f"/ {div}")
print(f"% {mod}")
```

Vectors

```r
grades <- c(90,88,96,83,91)

grades

cat("MEAN:", mean(grades))
cat("\nMAX:", max(grades))
cat("\nMIN:", min(grades))
cat("\nSUM:", sum(grades))
cat("\nLENGTH:", length(grades))
```
