# 1. Create the Student Database

```r
students <- data.frame(
    StudentID = c("S001", "S002", "S003", "S004", "S005", "S006", "S007", "S008"),
    Programming_Grade =c(88, 72, 91, 78, 65, 85, 74, 82),
    Math_Grade = c(85, 70, 88, 80, 68, 90, 73, 79),
    Attendance = c(92, 80, 95, 88, 70, 91, 76, 85),
    Prerequisite = c("Yes", "No", "Yes", "Yes", "No", "Yes", "No", "Yes")
)
write.csv(students, "Students.csv")

students
```

7. Implement the Rules

```r
eligibility <- function(programming, math, att, prq) {
  if (programming >= 80 && math >= 75 && att >= 80 && prq == "Yes"){
    return("Eligible")
  } else if (programming >= 75 && math >= 70 && att >= 75){
    return("Conditional")
  } else {
    return("Not Eligible")
  }
}
```

8. Apply all the Function


```r
students$Status <- mapply(
  eligibility,
  students$Programming_Grade,
  students$Math_Grade,
  students$Attendance,
  students$Prerequisite
)

students
```

9. Recommendation

```r
student_Recommendation <- function(eligibility){
  if(eligibility == "Eligible"){
    return("Student may enroll")
  }else if (eligibility == "Conditional") {
    return("Consult academic yutor")
  }else {
    return("Complete requirements before enrollment.")
  }
}
student_Recommendation("Eligible")
```

10. Complete the table

```r
students$Recomendation <- mapply(
  student_Recommendation,
  students$Status
)

students
```
